# Test połączenia z serwerem prowadzącego

Ten notebook sprawdza czy możesz połączyć się z LLM-em hostowanym przez prowadzącego.

**Przed uruchomieniem:** upewnij się że jesteś w tej samej sieci WiFi co prowadzący.

Prowadzący poda **adres** (`http://ADRES:PORT`) i **hasło**. Po uruchomieniu komórki zostaniesz zapytany o imię (widoczne u prowadzącego) i — jeśli serwer tego wymaga — o hasło.

```python
from utils import connect_llm, setup_auth_client

# === WPISZ ADRES PODANY PRZEZ PROWADZĄCEGO ===
LECTURER_SERVER = "http://ADRES_SERWERA:PORT"
# ==============================================

client, _, MODEL_NAME = connect_llm(lecturer_server=LECTURER_SERVER, backend="lecturer")
client, _ = setup_auth_client(client, None, MODEL_NAME)

# === Test chatu ===
if client:
    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role": "user", "content": "Powiedz cześć po polsku w jednym zdaniu."}],
        max_tokens=60,
    )
    print(f"Model: {MODEL_NAME}")
    print(f"Model odpowiedział: {response.choices[0].message.content}")
    print("\n✓ Wszystko działa! Możesz przejść do notebooka z zajęć.")
else:
    print("✗ Brak połączenia. Sprawdź adres, sieć WiFi i hasło.")
```
